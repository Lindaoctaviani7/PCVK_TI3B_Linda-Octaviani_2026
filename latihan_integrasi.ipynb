```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

STEP 2

```python
# membuat list untuk menyimpan url dari beberapa image
urls = [
    "https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
    "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
]
#baca dan tampilkan image
#loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                            #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)   #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)    #convert color to RGB
  final_frame = cv.hconcat((image, image_2))        #concatenate image
  cv2_imshow(final_frame)                           #show image
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi," x ",lebar)
cv2_imshow(image_2)
```

```python
image_gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
image_gray_display = cv.cvtColor(image_gray, cv.COLOR_GRAY2BGR)
image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for x in range(lebar):
  image_1[tinggi//2, x] = [255,255,255]

final_frame = cv.hconcat((image_gray_display, image_1))
cv2_imshow(final_frame)
```

```python
image_gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
image_gray_display = cv.cvtColor(image_gray, cv.COLOR_GRAY2BGR)
image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

cv.line(image_1, (0, 0), (lebar - 1, tinggi - 1),
        (255,255,255), 2)
cv.line(image_1, (lebar - 1, 0), (0, tinggi - 1),
        (255,255,255), 2)
hasil = cv.hconcat([image_gray_display, image_1])

cv2_imshow(hasil)
```

GARIS HORIZONTAL PUTIH DITENGAH

```python
image_gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
image_gray_display = cv.cvtColor(image_gray, cv.COLOR_GRAY2BGR)
image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# mengambil ukuran gambar
tinggi, lebar = image_1.shape[:2]

# membuat garis horizontal kecil di tengah image
panjang_garis = 80  # ukuran panjang garis

x_awal = (lebar//2) - (panjang_garis//2)
x_akhir = (lebar//2) + (panjang_garis//2)

for x in range(x_awal, x_akhir):
    image_1[tinggi//2, x] = [255,255,255]

final_frame = cv.hconcat((image_gray_display, image_1))
cv2_imshow(final_frame)
```

MEMBUAT KOTAK KUMPULAN PIXEL

```python
image_gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
image_gray_display = cv.cvtColor(image_gray, cv.COLOR_GRAY2BGR)
image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Mengambil ukuran gambar
tinggi, lebar = image_1.shape[:2]

# Ukuran kotak kecil di tengah
ukuran_kotak = 80   # panjang sisi kotak pixel

# Menentukan koordinat kotak
x_awal = (lebar//2) - (ukuran_kotak//2)
x_akhir = (lebar//2) + (ukuran_kotak//2)

y_awal = (tinggi//2) - (ukuran_kotak//2)
y_akhir = (tinggi//2) + (ukuran_kotak//2)

# Mengubah kumpulan pixel menjadi warna putih
for y in range(y_awal, y_akhir):
    for x in range(x_awal, x_akhir):
        image_1[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_gray_display, image_1))
cv2_imshow(final_frame)
```

8. Buat rectangle pada bagian sudut bawah kiri channel B pada color space
RGB dari citra kitten/ lena/ mandrill/ male/ female/ couple/ sailboat/
peppers!  

```python
import cv2
import matplotlib.pyplot as plt

image_path = "/content/drive/MyDrive/Data Saya/semester 5/pengelolahan citra dan visi komputer/ktm.jpeg" # Placeholder: Please update this path

# Load the image
img_target = cv2.imread(image_path)

# Check if the image was loaded successfully
if img_target is None:
    print(f"Error: Could not load image from {image_path}. Please check the path.")
else:
    # Convert BGR to RGB for Matplotlib display
    img_rgb_target = cv2.cvtColor(img_target, cv2.COLOR_BGR2RGB)

    # Create a copy to work on, preserving the original
    modified_img = img_rgb_target.copy()

    # Get image dimensions
    h, w, _ = modified_img.shape

    # Define the coordinates for a rectangle in the bottom-left corner
    # You can adjust these coordinates as needed
    rect_start_x = int(w * 0.05)  # 5% from the left edge
    rect_start_y = int(h * 0.70)  # 70% from the top edge (making it bottom-ish)
    rect_end_x = int(w * 0.30)    # 30% from the left edge
    rect_end_y = int(h * 0.95)    # 95% from the top edge (closer to the bottom)

    # Set the blue channel (index 2) within the rectangle to maximum intensity (255)
    # The order for RGB is 0=Red, 1=Green, 2=Blue
    modified_img[rect_start_y:rect_end_y, rect_start_x:rect_end_x, 2] = 255

    # Display the result
    plt.figure(figsize=(10, 8))
    plt.imshow(modified_img)
    plt.title(f"Rectangle on Blue Channel of {image_path.split('/')[-1]} (Bottom-Left)")
    plt.axis("off")
    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

image_path = "/content/drive/MyDrive/Data Saya/semester 5/pengelolahan citra dan visi komputer/ktm.jpeg"

img = cv2.imread(image_path)

if img is None:
    print(f"Error: Could not load image from {image_path}. Please check the path.")
else:
    # Ubah BGR ke RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    modified_ktm = img_rgb.copy()

    height, width, _ = modified_ktm.shape

    nama_start_x = int(width * 0.215)
    nama_start_y = int(height * 0.755)

    nama_end_x = int(width * 0.515)
    nama_end_y = int(height * 0.815)

    cv2.rectangle(
        modified_ktm,
        (nama_start_x, nama_start_y),
        (nama_end_x, nama_end_y),
        (0, 255, 0),
        -1
    )

    nim_start_x = int(width * 0.215)
    nim_start_y = int(height * 0.825)

    nim_end_x = int(width * 0.395)
    nim_end_y = int(height * 0.865)

    cv2.rectangle(
        modified_ktm,
        (nim_start_x, nim_start_y),
        (nim_end_x, nim_end_y),
        (255, 0, 0),
        -1
    )

    prodi_start_x = int(width * 0.215)
    prodi_start_y = int(height * 0.875)

    prodi_end_x = int(width * 0.465)
    prodi_end_y = int(height * 0.925)

    cv2.rectangle(
        modified_ktm,
        (prodi_start_x, prodi_start_y),
        (prodi_end_x, prodi_end_y),
        (0, 0, 255),
        -1
    )

    plt.figure(figsize=(12, 8))
    plt.imshow(modified_ktm)
    plt.title("KTM dengan NIM, Nama, dan Prodi Ditutup")
    plt.axis("off")
    plt.show()
```
