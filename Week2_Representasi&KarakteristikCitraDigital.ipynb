1. Perbedaan tampilan terjadi karena OpenCV menggunakan format BGR, sedangkan `plt.imshow()` menggunakan RGB. Akibatnya, kanal merah dan biru tertukar sehingga warna terlihat berbeda, meskipun nilai piksel tetap sama. Kesimpulan: perbedaan warna disebabkan oleh perbedaan urutan kanal BGR dan RGB, bukan perubahan nilai piksel.


```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# Baca citra
img = cv2.imread('/content/drive/MyDrive/Data Saya/semester 5/pengelolahan citra dan visi komputer/meong.jpeg')

# Tampilkan tanpa konversi warna
cv2_imshow(img)

plt.figure(figsize=(8, 5))
plt.imshow(img)
plt.title('Citra menggunakan plt.imshow() tanpa konversi')
plt.axis('off')
plt.show()

# Ambil satu piksel yang sama
y, x = 100, 100

print("Nilai piksel dari img[y, x] :", img[y, x])
print("B = ", img[y, x][0])
print("G = ", img[y, x][1])
print("R = ", img[y, x][2])
```

2. Kedua citra tampak hitam karena semua nilai piksel bernilai 0, sehingga nilai min() dan max() juga sama.
Perbedaannya terletak pada penggunaan memori: int16 membutuhkan 2 byte, sedangkan int32 membutuhkan 4 byte per elemen. Jadi, int32 menggunakan memori dua kali lebih besar.

```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow

# Ukuran kanvas
tinggi = 500
lebar = 800

# Kanvas int16
img_int16 = np.zeros((tinggi, lebar, 3), dtype=np.int16)

# Kanvas int32
img_int32 = np.zeros((tinggi, lebar, 3), dtype=np.int32)

print("=== INT16 ===")
print("dtype :", img_int16.dtype)
print("min   :", img_int16.min())
print("max   :", img_int16.max())
print("nbytes:", img_int16.nbytes)

print("\n=== INT32 ===")
print("dtype :", img_int32.dtype)
print("min   :", img_int32.min())
print("max   :", img_int32.max())
print("nbytes:", img_int32.nbytes)

# Tampilkan
print("\nTampilan int16:")
cv2_imshow(img_int16)

print("Tampilan int32:")
cv2_imshow(img_int32)
```

4. skimage.io.imread() dan cv2.cvtColor(...BGR2RGB) sama-sama menghasilkan urutan RGB yang benar. Yang kanalnya tertukar adalah citra cv2.imread() apabila langsung diberikan ke plt.imshow().

```python
import cv2
import matplotlib.pyplot as plt
from skimage import io

path = '/content/drive/MyDrive/Data Saya/semester 5/pengelolahan citra dan visi komputer/meong.jpeg'

# Baca menggunakan OpenCV
img_bgr = cv2.imread(path)

# Konversi BGR -> RGB
img_rgb_cv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Baca menggunakan scikit-image
img_rgb_skimage = io.imread(path)

# Tampilkan
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb_skimage)
plt.title('skimage.io.imread()')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_rgb_cv)
plt.title('cv2.cvtColor(BGR → RGB)')
plt.axis('off')

plt.show()

# Periksa nilai piksel
y, x = 100, 100

print("Piksel skimage       :", img_rgb_skimage[y, x])
print("Piksel OpenCV RGB    :", img_rgb_cv[y, x])
print("Piksel OpenCV BGR    :", img_bgr[y, x])
```

1. Kombinasi Kanal Merah-Biru dan Hijau-Biru

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/drive/MyDrive/Data Saya/semester 5/pengelolahan citra dan visi komputer/meong.jpeg')

# Salinan citra untuk kombinasi kanal
merah_biru = img.copy()
hijau_biru = img.copy()

# Merah-Biru: kanal Hijau dinolkan
merah_biru[:, :, 1] = 0

# Hijau-Biru: kanal Merah dinolkan
hijau_biru[:, :, 2] = 0

# Konversi ke RGB untuk Matplotlib
merah_biru_rgb = cv2.cvtColor(merah_biru, cv2.COLOR_BGR2RGB)
hijau_biru_rgb = cv2.cvtColor(hijau_biru, cv2.COLOR_BGR2RGB)

# Tampilkan berdampingan
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(merah_biru_rgb)
plt.title('Kombinasi Kanal Merah + Biru')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(hijau_biru_rgb)
plt.title('Kombinasi Kanal Hijau + Biru')
plt.axis('off')

plt.tight_layout()
plt.show()
```

2. Tiga Bentuk Pencerminan
Modul menetapkan:
flipCode = 0 = vertikal/atas-bawah;
flipCode = 1 = horizontal/kiri-kanan;
flipCode = -1 = vertikal dan horizontal sekaligus.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/drive/MyDrive/Data Saya/semester 5/pengelolahan citra dan visi komputer/meong.jpeg')

# Pencerminan
flip_vertikal = cv2.flip(img, 0)
flip_horizontal = cv2.flip(img, 1)
flip_keduanya = cv2.flip(img, -1)

# Konversi RGB untuk matplotlib
flip_vertikal = cv2.cvtColor(flip_vertikal, cv2.COLOR_BGR2RGB)
flip_horizontal = cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB)
flip_keduanya = cv2.cvtColor(flip_keduanya, cv2.COLOR_BGR2RGB)

# Tampilkan
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertikal)
plt.title('Pencerminan Vertikal')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title('Pencerminan Horizontal')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_keduanya)
plt.title('Pencerminan Keduanya')
plt.axis('off')

plt.tight_layout()
plt.show()
```
