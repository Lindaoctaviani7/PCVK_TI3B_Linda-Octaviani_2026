```python
img_gray = cv.imread('WhatsApp Image 2026-09-04 at 13.21.49.jpeg', cv.IMREAD_GRAYSCALE)

#Tanpa cmap
plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title('Tanpa cmap')

#Dengan cmap
plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.title('Dengan cmap')

plt.show()
```

bentuk Geometri 2D dari OpenCV. Diawali dengan pembuatan black image dengan tipe data int16.

```python
black_img = np.zeros((530,530,3), dtype=np.int16)
plt.imshow(black_img)
```

menambahkan bentuk persegi sesuai koordinat pt1 dan pt2

```python
#perhatikan koordinator titik2 pt1 dan pt2
cv.rectangle(black_img,pt1=(290,0), pt2=(520,130),color=(0,0,255), thickness=10)
plt.imshow(black_img)
```

menambahkan bentuk persegi sesuai koordinat pt1 dan pt2

```python
cv.rectangle(black_img,pt1=(250,250), pt2=(350,350),color=(255,0,0), thickness=15)
plt.imshow(black_img)
```

menambahkan bentuk lingkaran sesuai radius

```python
cv.circle(black_img, center=(150,150), radius=60, color=(0,255,0), thickness=8)
plt.imshow(black_img)
```

penambahan lingkaran berwarna dan garis sesuai koordinat pt1 dan pt2

```python
cv.circle(black_img, center=(400,400), radius=50, color=(0,255,0), thickness=1)
cv.line(black_img, pt1=(0,0), pt2=(530,530), color=(255,255,255), thickness=5)
plt.imshow(black_img)
```

Penambahan text dengan font yang telah tertulis dengan ukuran yang sudah ditentukan.

```python
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text='PVCK is fun!',org=(10,500), fontFace=font, fontScale=2, color=(255,255,0), thickness=3, lineType=cv.LINE_AA)
plt.imshow(black_img)
```

Pembuatan black image kembali dilakukan dengan tipe data int32

```python
black_img2=np.zeros(shape=(530,530,3),dtype=np.int32)
plt.imshow(black_img2)
```

kode program untuk inisialisasi NumPy array dengan tipe data int32

```python
vertices = np.array([[150,350],[250,250],[450,350],[250,450]], dtype=np.int32)
vertices

pts = vertices.reshape((-1,1,2)) #nilai 2 untuk menunjukkab bahwa tiap titik dibuat 3 channel yang mewakili R, G, B
pts

#array([[150, 350],
     #  [250, 250],
      # [450, 350],
       #[250, 450]], dtype=int32)
```

Penambahan polyline pada black image kedua

```python
cv.polylines(black_img2, [pts], isClosed=True, color=(0,255,0), thickness=5)
plt.imshow(black_img2)
#
```
